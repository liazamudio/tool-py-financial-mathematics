# Tiempo y tasa

```python
# FUNCIONES

# Calcular el valor futuro tomando en cuenta cuantas veces se quiere incrementar el Valor Presente
def calValorFuturo(valorPresente,vecesIncrem):
  res_valorFuturo = valorPresente * vecesIncrem
  return res_valorFuturo

# Determinar el plazo
def determinaPlazo(valorFuturo, valorPresente, tax_int):
  floInt = taxInt/100
  resPlazo = round((((valorFuturo / valorPresente) - 1) / floInt), 2)
  return resPlazo

# Determinar la tasa de interes fija
def determinaInteres(valorFuturo, valorPresente, plazoInv):
  resInteres = round(((((valorFuturo / valorPresente) - 1) / plazoInv)*100), 2)
  return resInteres

# Determinar el valor futuro con tasa de interés fija
def determinaVFif(valorPresente, taxInt, plazoInv):
  floInt = taxInt/100
  resVFif = round((valorPresente * (1 + (floInt * plazoInv))), 2)
  return resVFif

# Determinar el valor futuro con tasa de interés cambiante - optimizar
def determinaVFiv3a(valorPresente, taxInt1, taxInt2, taxInt3):
  floIntAcum = (taxInt1 + taxInt2 + taxInt3)/100
  resVFiv = round((valorPresente * (1 + floIntAcum)), 2)
  return resVFiv

# Determinar el valor presente con tasa de interés fija
def determinaVPif(valorFuturo, taxInt, plazoInv):
  floInt = taxInt/100
  resVPif = round((valorFuturo / (1 + (floInt * plazoInv))), 2)
  return resVPif
```

¿Cuánto tiempo tendré que dejar una inversión para triplicar su valor si existe una tasa de interés en el mercado con la que puedo ganar del 1.9% mensual?

```python
# Definición de variables
tasa_mensual = 0.019  # 1.9% expresado en decimal
objetivo_multiplo = 3 # Queremos triplicar (3x)

# Cálculo del tiempo (n) usando logaritmos
# n = log(VF/VP) / log(1 + i)
# Como VF/VP es 3, simplemente usamos log(3)
meses = math.log(objetivo_multiplo) / math.log(1 + tasa_mensual)

# Conversión a años para mayor claridad
anios = meses / 12

print(f"Resultados de la Inversión:")
print(f"---------------------------")
print(f"Tiempo necesario en meses: {meses:.2f}")
print(f"Tiempo necesario en años:  {anios:.2f}")
```

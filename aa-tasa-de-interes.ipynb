# Tasa de Interes Total

En el mundo financiero, esta fórmula suele referirse a la descomposición de la tasa de interés:

- <b>RC: Riesgo Crediticio (Spread por impago)</b>. Es el riesgo que tenemos a que no nos pague.
- <b>RL: Riesgo de Liquidez</b>. Que tan fácil es vender esta deuda para el emisor, el comprador de la deuda, el inversionista.
- <b>PT: Prima Temporal (o Plazo)</b>. Dependiendo el tiempo que dure, es el premio que vamos a recibir. A mayor tiempo, mayor incertidumbre y por eso a mayor plazo, mayor tasa.

```python
# Formula
def calcular_tit(riesgo_crediticio: float, riesgo_liquidez: float, prima_temporal: float) -> float:
    """
    Calcula la Tasa de Interés Total (TIT) basada en componentes de riesgo.
    
    Args:
        riesgo_crediticio (float): Spread por riesgo de crédito (ej. 0.02 para 2%).
        riesgo_liquidez (float): Prima por liquidez.
        prima_temporal (float): Prima por el plazo del activo.
        
    Returns:
        float: Tasa de Interés Total resultante.
    """
    return riesgo_crediticio + riesgo_liquidez + prima_temporal

# Ejemplo de uso:
rc = 0.025  # 2.5%
rl = 0.010  # 1.0%
pt = 0.015  # 1.5%

tit = calcular_tit(rc, rl, pt)

print(f"La Tasa de Interés Total es: {tit:.2%}")
```

# Tasa de interes efectiva

Una tasa de interés efectiva es lo que realmente ganas por cada inversión o lo que pagas por cada crédito si lo mantienes al vencimiento. DATO IMPORTANTE, EN CUALQUIER EJERCICIO SIEMPRE ADAPTADA TASA A PERIODICIDAD (Mensual, Bimestral, Cuatrimestral, Trimestral, Semestral, Anual, Diaria) 

## Interes simple

```python
# Tasa de interes simple
def calcular_tasa_simple(interes_total: float, principal: float, tiempo: float) -> float:
    """
    Calcula la tasa de interés simple anual.
    
    Args:
        interes_total (float): Monto total de intereses en dinero.
        principal (float): Capital inicial.
        tiempo (float): Tiempo en años.
        
    Returns:
        float: Tasa de interés en formato decimal.
    """
    if principal <= 0 or tiempo <= 0:
        raise ValueError("El principal y el tiempo deben ser mayores a cero.")
        
    tasa = interes_total / (principal * tiempo)
    return tasa

```

Ejemplo:

```python
# Ganaste $150 de interés sobre un capital de $1,000 en 2 años.
I = 150
P = 1000
T = 2

r_simple = calcular_tasa_simple(I, P, T)

print(f"La tasa de interés simple anual es: {r_simple:.2%}")
```

## Tasa de interes compuesto

```python
import numpy as np

def calcular_tasa_compuesta(valor_futuro: float, principal: float, periodos: int) -> float:
    """
    Calcula la tasa de interés compuesto por periodo.
    
    Args:
        valor_futuro (float): Monto final obtenido.
        principal (float): Capital invertido inicialmente.
        periodos (int): Número de periodos (años, meses, etc.).
        
    Returns:
        float: Tasa de interés decimal por periodo.
    """
    if principal <= 0 or valor_futuro <= 0:
        raise ValueError("Los montos deben ser positivos.")
    
    # Aplicamos la fórmula despejada: r = (VF/P)^(1/n) - 1
    tasa = (valor_futuro / principal) ** (1 / periodos) - 1
    return tasa
```

Ejemplo:

```python
# Invertiste $1,000 y tras 5 años tienes $1,610.51
vf = 1610.51
p = 1000
n = 5

tasa_anual = calcular_tasa_compuesta(vf, p, n)
print(f"La tasa de interés compuesto anual es: {tasa_anual:.2%}")
```

## Tasa de interés real

Una tasa de interés real es una tasa de interés que <b>se ha ajustado para eliminar los efectos de la inflación y reflejar el costo real de los fondos para el prestatario y el rendimiento real para el prestamista o el inversionista</b>. La tasa de interés real refleja la tasa de preferencia temporal de los bienes actuales sobre los futuros.

```python
def calcular_tasa_real(tasa_nominal: float, inflacion: float) -> float:
    """
    Calcula la tasa de interés real exacta (Ecuación de Fisher).
    
    Args:
        tasa_nominal (float): Tasa de interés nominal (ej. 0.08 para 8%).
        inflacion (float): Tasa de inflación (ej. 0.03 para 3%).
        
    Returns:
        float: Tasa de interés real.
    """
    # r = ((1 + i) / (1 + pi)) - 1
    tasa_real = ((1 + tasa_nominal) / (1 + inflacion)) - 1
    return tasa_real
```

Ejemplo:

```python
i_nom = 0.10  # El banco te da el 10%
infla = 0.04  # La inflación fue del 4%

r_real = calcular_tasa_real(i_nom, infla)

print(f"Tasa Nominal: {i_nom:.2%}")
print(f"Inflación:    {infla:.2%}")
print(f"Tasa Real:     {r_real:.2%}")
```

# Interés simple

En el interés simple, los intereses nunca se reinvierten, nunca pasan a formar parte del capital y lo único que genera valor es la inversión inicial.

```python
# Funciones para interés simple

# Agregar las incluidas en diversoso ejercicios de esta notebook

# ...

# Calcular el valor futuro tomando en cuenta cuantas veces se quiere incrementar el Valor Presente
def calValorFuturo(valorPresente,vecesIncrem):
  res_valorFuturo = valorPresente * vecesIncrem
  return res_valorFuturo

# Determinar el plazo
def determinaPlazo(valorFuturo, valorPresente, tax_int):
  floInt = taxInt/100
  resPlazo = round((((valorFuturo / valorPresente) - 1) / floInt), 2)
  return resPlazo

# Determinar la tasa de interes fija
def determinaInteres(valorFuturo, valorPresente, plazoInv):
  resInteres = round(((((valorFuturo / valorPresente) - 1) / plazoInv)*100), 2)
  return resInteres

# Determinar el valor futuro con tasa de interés fija
def determinaVFif(valorPresente, taxInt, plazoInv):
  floInt = taxInt/100
  resVFif = round((valorPresente * (1 + (floInt * plazoInv))), 2)
  return resVFif

# Determinar el valor futuro con tasa de interés cambiante - optimizar
def determinaVFiv3a(valorPresente, taxInt1, taxInt2, taxInt3):
  floIntAcum = (taxInt1 + taxInt2 + taxInt3)/100
  resVFiv = round((valorPresente * (1 + floIntAcum)), 2)
  return resVFiv

# Determinar el valor presente con tasa de interés fija
def determinaVPif(valorFuturo, taxInt, plazoInv):
  floInt = taxInt/100
  resVPif = round((valorFuturo / (1 + (floInt * plazoInv))), 2)
  return resVPif


```

## Tiempo y tasa

### Formulas

```python
# Funciones para interés simple
def calcular_interes_simple(vp, vf, n):
    """Calcula la tasa de interés (i) dado VP, VF y tiempo."""
    # Formula: i = ((VF / VP) - 1) / n
    return ((vf / vp) - 1) / n

def calcular_tiempo_simple(vp, vf, i):
    """Calcula el tiempo (n) dado VP, VF y tasa de interés."""
    # Formula: n = ((VF / VP) - 1) / i
    return ((vf / vp) - 1) / i

def calcular_valor_presente_simple(vf, i, n):
    """Calcula el Valor Presente (VP) dado VF, tasa y tiempo."""
    # Formula: VP = VF / (1 + i * n)
    return vf / (1 + (i * n))

def calcular_valor_futuro_simple(vp, i, n):
    """Calcula el Valor Futuro (VF) dado VP, tasa y tiempo."""
    # Formula: VF = VP * (1 + i * n)
    return vp * (1 + (i * n))
```

## Valor futuro

El valor futuro (FV) es el valor de un activo corriente en una fecha futura basado en una tasa de interés supuesta. <b>El valor futuro (FV) es importante para los inversores y los planificadores financieros, ya que lo utilizan para estimar cuánto valdrá una inversión realizada hoy en el futuro.</b>

### Valor futuro utilizando el interés simple

```python
# import pandas as pd

def calcular_valor_futuro_simple_anualizado(pv: float, i_anual: float, t: float, unidad: str = 'años') -> float:
    """
    Calcula el Valor Futuro usando interés simple con una tasa inicial anualizada.
    
    Args:
        pv (float): Valor Presente o Capital Inicial.
        i_anual (float): Tasa de interés anual en formato decimal (ej. 0.12 para 12%).
        t (float): Plazo o número de periodos.
        unidad (str): 'años' o 'meses'. Determina cómo se ajusta la tasa al tiempo.
        
    Returns:
        float: Valor Futuro (FV).
    """
    unidad = unidad.lower()
    
    if unidad == 'años':
        # El tiempo ya es anual, se usa directo
        t_ajustado = t
    elif unidad == 'meses':
        # Se convierte el tiempo de meses a fracción de año
        t_ajustado = t / 12
    else:
        raise ValueError("La unidad debe ser 'años' o 'meses'.")
        
    return pv * (1 + (i_anual * t_ajustado))
```

Supongamos una inversión de $100,000 a una tasa del 18% anual durante 2 años.

```python
pv = 100000
tasa_anual = 0.18
plazo_anios = 2

fv_anios = calcular_valor_futuro_simple_anualizado(pv, tasa_anual, plazo_anios, unidad='años')

print(f"--- Caso Anual ---")
print(f"Inversión: ${pv:,.2f} | Tasa: {tasa_anual:.2%} anual | Plazo: {plazo_anios} años")
print(f"Valor Futuro: ${fv_anios:,.2f}")
```

### Tasa de interés cambiante en el tiempo.

Lo que hemos visto hasta el momento solo funciona si la tasa es fija a través del tiempo, per considerando que las tasas de interese son cambiantes, debemos entender el proceso para calcular el valor futuro.

La formula general mantendrá una estructura similar donde lo que haremos es multiplicar el valor presente por uno mas la tasa efectiva, pero en este caso la tasa efectiva cambiara al ser una suma de las tasas de interés para cada periodo:

𝐹𝑉=𝑃𝑉∗(1+(𝑖1+𝑖2+⋯+𝑖𝑛))

Aplicado en Python

```python
# Función para calcular el Valor Futuro con tasas variables
def calcular_vf_tasas_variables(pv: float, tasas: list) -> float:
    """
    Calcula el Valor Futuro con Interés Simple y tasas variables.
    Fórmula: FV = PV * (1 + sum(tasas))
    
    Args:
        pv (float): Valor Presente.
        tasas (list): Lista de tasas decimales por periodo (ej. [0.01, 0.012, 0.009]).
    """
    suma_tasas = sum(tasas)
    return pv * (1 + suma_tasas)
```

## Valor Presente

El valor presente (PV) es el valor actual de una suma futura de dinero o flujo de flujos de efectivo dada una tasa de rendimiento específica Los flujos de efectivo futuros se descuentan a la tasa de descuento, y cuanto mayor sea la tasa de descuento, menor será el valor presente de los flujos de efectivo futuros Determinar la tasa de descuento adecuada es la clave para valorar adecuadamente los flujos de efectivo futuros, ya sean ganancias u obligaciones.

```python
def calcular_valor_presente_simple(fv: float, i_anual: float, t: float, unidad: str = 'años') -> float:
    """
    Calcula el Valor Presente (PV) a partir de una tasa anualizada.
    Fórmula: PV = FV / (1 + (i * t))
    """
    unidad = unidad.lower()
    
    if unidad == 'años':
        t_ajustado = t
    elif unidad == 'meses':
        # Se divide el tiempo entre 12 para que sea proporcional a la tasa anual
        t_ajustado = t / 12
    else:
        raise ValueError("La unidad debe ser 'años' o 'meses'.")
        
    return fv / (1 + (i_anual * t_ajustado))
```

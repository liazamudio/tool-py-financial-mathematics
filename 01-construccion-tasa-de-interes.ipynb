# Construcción de la Tasa de Interés

Fórmulas base para construir y derivar tasas de interés: total (por componentes de riesgo), simple, compuesta, por periodicidad, efectiva de un préstamo y real.

```python
import pandas as pd  # Para manejo de tablas y series temporales
import numpy as np   # Para cálculos matemáticos optimizados
```

## 1. Tasa de Interés Total (TIT)

La TIT se construye sumando el spread por riesgo de crédito, la prima por liquidez y la prima por el plazo del activo.

```python
# Formula
def calcular_tit(riesgo_crediticio: float, riesgo_liquidez: float, prima_temporal: float) -> float:
    """
    Calcula la Tasa de Interés Total (TIT) basada en componentes de riesgo.

    Args:
        riesgo_crediticio (float): Spread por riesgo de crédito (ej. 0.02 para 2%).
        riesgo_liquidez (float): Prima por liquidez.
        prima_temporal (float): Prima por el plazo del activo.

    Returns:
        float: Tasa de Interés Total resultante.
    """
    return riesgo_crediticio + riesgo_liquidez + prima_temporal
```

```python
# Ejemplo de uso:
rc = 0.025  # 2.5%
rl = 0.010  # 1.0%
pt = 0.015  # 1.5%

tit = calcular_tit(rc, rl, pt)

print(f"La Tasa de Interés Total es: {tit:.2%}")
```

## 2. Tasa de interés simple (a partir del interés total)

Despeje de la tasa anual conociendo el interés total generado, el principal y el tiempo.

```python
# Tasa de interes simple
def calcular_tasa_simple(interes_total: float, principal: float, tiempo: float) -> float:
    """
    Calcula la tasa de interés simple anual.

    Args:
        interes_total (float): Monto total de intereses en dinero.
        principal (float): Capital inicial.
        tiempo (float): Tiempo en años.

    Returns:
        float: Tasa de interés en formato decimal.
    """
    if principal <= 0 or tiempo <= 0:
        raise ValueError("El principal y el tiempo deben ser mayores a cero.")

    tasa = interes_total / (principal * tiempo)
    return tasa
```

```python
# Ganaste $150 de interés sobre un capital de $1,000 en 2 años.
I = 150
P = 1000
T = 2

r_simple = calcular_tasa_simple(I, P, T)

print(f"La tasa de interés simple anual es: {r_simple:.2%}")
```

## 3. Tasa de interés compuesta (a partir del valor futuro)

Despeje de la tasa periódica conociendo el valor futuro, el principal y el número de periodos.

```python
def calcular_tasa_compuesta(valor_futuro: float, principal: float, periodos: int) -> float:
    """
    Calcula la tasa de interés compuesto por periodo.

    Args:
        valor_futuro (float): Monto final obtenido.
        principal (float): Capital invertido inicialmente.
        periodos (int): Número de periodos (años, meses, etc.).

    Returns:
        float: Tasa de interés decimal por periodo.
    """
    if principal <= 0 or valor_futuro <= 0:
        raise ValueError("Los montos deben ser positivos.")

    # Aplicamos la fórmula despejada: r = (VF/P)^(1/n) - 1
    tasa = (valor_futuro / principal) ** (1 / periodos) - 1
    return tasa
```

```python
# Invertiste $1,000 y tras 5 años tienes $1,610.51
vf = 1610.51
p = 1000
n = 5

tasa_anual = calcular_tasa_compuesta(vf, p, n)
print(f"La tasa de interés compuesto anual es: {tasa_anual:.2%}")
```

## 4. Tasa de interés efectiva por periodicidad

Determinar la tasa mensual, bimestral, trimestral, cuatrimestral y semestral a partir de una tasa anual dada.

```python
# DEFINICIÓN DE FUNCIONES

# Tasa anual
def tax_anual(tasa_anual):
  t_anual = round((tasa_anual),2)
  return t_anual

# Tasa mensual
def tax_mensual(tasa_anual):
  t_mensual = round((tasa_anual/12),2)
  return t_mensual

# Tasa bimestral
def tax_bimestral(tasa_anual):
  t_bimestral = round((tasa_anual/6),2)
  return t_bimestral

# Tasa trimestral
def tax_trimestral(tasa_anual):
  t_trimestral = round((tasa_anual/4),2)
  return t_trimestral

# Tasa cuatrimestral
def tax_cuatrimestral(tasa_anual):
  t_cuatrimestral = round((tasa_anual/3),2)
  return t_cuatrimestral

# tasa semestral
def tax_semestral(tasa_anual):
  t_semestral = round((tasa_anual/2),2)
  return t_semestral
```

```python
# Ejemplo: tasa anual del 12%
tasa_anual = 12.0

tax_anu = tax_anual(tasa_anual)
tax_men = tax_mensual(tasa_anual)
tax_bim = tax_bimestral(tasa_anual)
tax_trm = tax_trimestral(tasa_anual)
tax_ctm = tax_cuatrimestral(tasa_anual)
tax_sem = tax_semestral(tasa_anual)

print(' ')
print(f'La tasa de interes anual es del: {tax_anu} %.')
print(f'Mensual: {tax_men}%.')
print(f'Bimestral: {tax_bim}%.')
print(f'Trimestral: {tax_trm}%.')
print(f'Cuatrimestral: {tax_ctm}%.')
print(f'Semestral: {tax_sem}%.')
```

## 5. Tasa de Interés Efectiva (TIE) de un préstamo

Determinar cuál será la tasa de interés efectiva para un préstamo invertido a un plazo anual si no se capitalizan intereses, o si se capitalizan cada mes.

```python
#   DEFINICIÓN DE FUNCIONES

# Interés simple
def inter_simp(tasa_anual, plazo):
  res_int_simp = round((tasa_anual * plazo),2)
  return res_int_simp

# Interes compuesto
def inter_com(tasa_anual, plazo):
  tasa_mens = ((tasa_anual/12)/100)
  periodos = (plazo * 12)
  res_inter_comp = round((((pow((1+tasa_mens),periodos))-1)*100),2)
  return res_inter_comp
```

```python
# Ejemplo: tasa anual del 12% a 2 años
tasa_anual = 12.0
plazo = 2.0

int_sim = inter_simp(tasa_anual, plazo)
int_com = inter_com(tasa_anual, plazo)

print(' ')
print(f'Si invertimos a una tasa del {tasa_anual}% anual a {plazo} años:')
print(f'Si no se capitalizan los intereses obtenemos una TIE del {int_sim}%.')
print(f'Si se capitalizan los intereses cada mes obtenemos una TIE del {int_com}%.')
```

## 6. Tasa de interés real

¿Cuál será la tasa de interés real dado un rendimiento nominal y una inflación anual? Se calcula una versión aproximada (resta simple) y la versión exacta (Ecuación de Fisher).

```python
# DEFINICIÓN DE FUNCIONES

# Tasa real (aproximada)
def tax_real(por_rendimiento, por_inflacion):
  res_tasa_real = round((por_rendimiento - por_inflacion),2)
  return res_tasa_real

# Tasa de interes real (exacta, Ecuación de Fisher)
def taxin_real(por_rendimiento, por_inflacion):
  por_ren = (por_rendimiento / 100)
  por_inf = (por_inflacion / 100)
  res_tain_real = round(((((1+por_ren)/(1+por_inf))-1)*100),2)
  return res_tain_real
```

```python
# Ejemplo: rendimiento del 10% con una inflación del 4% anuales
por_rendimiento = 10.0
por_inflacion = 4.0

tasa_real = tax_real(por_rendimiento, por_inflacion)
tain_real = taxin_real(por_rendimiento, por_inflacion)

print(' ')
print(f'Con un rendimiento del {por_rendimiento}% y una inflación del {por_inflacion}% anuales, tenemos que:')
print(f'- Tenemos una tasa real anual del {tasa_real}%.')
print(f'- Y una tasa de interés real anual del {tain_real}%.')
```

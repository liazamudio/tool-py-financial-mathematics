# Tiempo y tasa

```python
# IMPORTACIONES
import math
```

```python
# FUNCIONES

# Calcula el valor futuro tomando en cuenta cuantas veces se quiere incrementar el Valor Presente
def fastFuturValue(presentValue,vecesIncrem):
  resFFuturValue = presentValue * vecesIncrem
  return resFFuturValue

# Calcula el tiempo
def calcTiempo(futureValue, presentValue, taxInt):
  floInt = taxInt/100
  resTime = (math.log(futureValue / presentValue)) / math.log(1 + floInt)
  return resTime

# Calcula la tasa de interés
def calcTasa(futureValue, presentValue,periods):
  resTax = ((pow((futureValue / presentValue),(1 / periods))) - 1) * 100
  return resTax

# calcula el Valor Futuro aplicando el interés compuesto con una tasa anual fija y reinvirtiendo los meses mensualmente
def calcFutureValueTF(presentValue, taxInt, periods):
  floIntMens = taxInt / (100*12)    # Mensualiza la tasa
  periodsMens = periods * 12        # Mensualiza el plazo
  resFutureValue = presentValue * pow((1+floIntMens), periodsMens)
  return resFutureValue

# calcula el Valor Futuro aplicando el interés compuesto con una tasa anual variable y reinvirtiendo los meses bimestralmente
def calcFutureValueTV(presentValue, taxInt1, taxInt2, taxInt3, taxInt4, taxInt5, periods):
  floTaxA1 = pow((1+((taxInt1/100)/periods)),periods)
  floTaxA2 = pow((1+((taxInt2/100)/periods)),periods)
  floTaxA3 = pow((1+((taxInt3/100)/periods)),periods)
  floTaxA4 = pow((1+((taxInt4/100)/periods)),periods)
  floTaxA5 = pow((1+((taxInt5/100)/periods)),periods)
  resFutureValue = presentValue * floTaxA1 * floTaxA2 * floTaxA3 * floTaxA4 * floTaxA5
  return resFutureValue
  # Optimizar usando un while (hasta que se cumpla el tiempo de la inversión) a la hora de solicitar los intereses por año

# Calcular el Valor Presente aplicando interes compuesto con una tasa fija de intereses en N cantidad de periodos de tiempo
def calcPresenteValueTF(futureValue, taxInt, periods, years):
  floTax = (taxInt/(100 * periods))   # Tasa quincenal
  totQui = periods * years
  resPresentValue = futureValue / (pow((1 + floTax),totQui))
  return resPresentValue

# Calcular el Valor Presente aplicando interes compuesto con una tasa variable de intereses en N cantidad de periodos de tiempo
def calcPresenteValueTV(futureValue, periods, years, taxInt1, taxInt2, taxInt3, taxInt4, taxInt5):
  floTaxA1 = pow((1+((taxInt1/100)/periods)),periods)
  floTaxA2 = pow((1+((taxInt2/100)/periods)),periods)
  floTaxA3 = pow((1+((taxInt3/100)/periods)),periods)
  floTaxA4 = pow((1+((taxInt4/100)/periods)),periods)
  floTaxA5 = pow((1+((taxInt5/100)/periods)),periods)
  resPresentValue = futureValue / (floTaxA1 * floTaxA2 * floTaxA3 * floTaxA4 * floTaxA5)
  return resPresentValue
  # Optimizar usando un while (hasta que se cumpla el tiempo de la inversión) a la hora de solicitar los intereses por año
```

¿Cuánto tiempo tendré que dejar una inversión para N veces su valor si existe una tasa de interés en el mercado con la que puedo ganar del xx.xx% mensual?

```python
# Introducimos la información
print('Introduce la siguiente información:\n')
presentValue= float(input("- Monto inicial de la inversión: "))
vecesIncrem = float(input("- Número de veces que quieres incrementar tu inversión: "))
taxInt = float(input("- Tasa de interes mensual: "))

# Lamamos a las funciones
futureValue = fastFuturValue(presentValue,vecesIncrem)
plazoFinal = calcTiempo(futureValue, presentValue, taxInt)

# Imprimimos resultados
# print(f'\n Si se invierten ${presentValue} y se quiere aumentar {vecesIncrem} veces su valor para tener ${futureValue} a una tasa d einterés mensual del {taxInt}%,')
print(f'\n se requiere de un plazo de {round(plazoFinal, 1)} meses para lograrlo')
```

¿Cuál es la tasa de interés que se requiere para que en N meses un valor de xx.xx invertidos se conviertan en xx.xx?

```python
# Introducimos la información
print('Introduce la siguiente información:\n')
presentValue= float(input("- Monto inicial de la inversión: "))
futureValue = float(input("- Monto final que se desea obtener: "))
periods = float(input("- Plazo de la inversión: "))

# Lamamos a las funciones
tasaReq = calcTasa(futureValue, presentValue,periods)

# Imprimimos el resultado
print(f'\n Se requiere una tasa del {round(tasaReq, 2)}% para lograrlo')
```

```python
# Introducimos la información
print('Introduce la siguiente información:\n')
presentValue= float(input("- Monto inicial de la inversión: "))
taxInt = float(input("- Tasa de interes mensual: "))
periods = float(input("- Plazo de la inversión en años: "))

# Lamamos a las funciones
futureValueTF = calcFutureValueTF(presentValue, taxInt, periods)

# Imprimimos el resultado
print(f'\n Se van a obtener ${round(futureValueTF, 2)}.')
```
